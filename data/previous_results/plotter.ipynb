```python
import numpy as np
import matplotlib.pyplot as plt

def load_data(name_file):
    data = np.loadtxt(name_file)
    # Extract columns
    p = data[:, 0]
    Pl = data[:, 1]
    
    return p, Pl

ng = 6
nbr = 10

# Load data from the file
#           BP+CB DECODING

name_file = f"data_l6_m6_ng6_nb10.txt"
name_file_2 = f"data_l6_m6.txt"
# name_file = f"data_l6_m6_ng6_nb10.txt"
p66, Pl66 = load_data(name_file)
p66_2, Pl66_2 = load_data(name_file_2)

name_file = "data_l9_m6_ng6_nb10.txt"
name_file_2 = f"data_l9_m6.txt"
# name_file = "data_l9_m6_ng6_nb10.txt"
p96, Pl96 = load_data(name_file)
p96_2, Pl96_2 = load_data(name_file_2)


name_file = "data_l12_m6_ng6_nb10.txt"
name_file_2 = f"data_l12_m6.txt"
# name_file = "data_l12_m6_ng6_nb10.txt"
p126_2, Pl126_2 = load_data(name_file_2)
p126, Pl126 = load_data(name_file)

#           BP+OSD DECODING

name_file = "bposd_data_l6_m6.txt"
p66bposd, Pl66bposd = load_data(name_file)


name_file = "bposd_data_l9_m6.txt"
p96bposd, Pl96bposd = load_data(name_file)

name_file = "bposd_data_l12_m6.txt"
p126bposd, Pl126bposd = load_data(name_file)



from mpl_toolkits.axes_grid1.inset_locator import inset_axes
# Plot in log-log scale
# plt.plot(p66_2, Pl66_2, marker='v', color = 'red', linestyle='', label='[[72, 12, 6]] BP+CB')
plt.plot(p66bposd, Pl66bposd, marker='+', color = 'red', linestyle='', label='[[72, 12, 6]] BPOSD')
plt.plot([min(p66), max(p66)], [min(p66), max(p66)], color='purple', linestyle='--', label='$P_L= p$')
# plt.plot(p96_2, Pl96_2, marker='v',color = 'green', linestyle='', label='[[108, 8, 10]] BP+CB')
plt.plot(p96bposd, Pl96bposd, marker='+', color = 'green', linestyle='', label='[[108, 8, 10]] BPOSD')
# plt.plot(p126_2, Pl126_2, marker='v', color = 'orange', linestyle='', label='[[144, 12, 12]] BP+CB')
plt.plot(p126bposd, Pl126bposd, marker='+', color = 'orange', linestyle='', label='[[144, 12, 12]] BPOSD')
plt.xlabel('$p$')
plt.ylabel('$P_L$')
# plt.title(f'ng = {ng}, nbr = {nbr}')
plt.xscale('log')
plt.yscale('log')


plt.legend(bbox_to_anchor=(0.5, -0.45), loc='lower center', ncol=3)
plt.margins(0.1, .1)

# 
plt.savefig("cbbposddata.pdf", format="pdf", bbox_inches='tight')
plt.show()

```
