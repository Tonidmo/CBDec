```python
from color_code_stim.color_code_stim import ColorCode
from beliefmatching import detector_error_model_to_check_matrices
import numpy as np
from scipy.io import savemat
```

```python
d = 11
p = 1e-3
rounds = d

code = ColorCode(
    d = d,
    rounds = d,
    p_bitflip= p,
    p_reset = p,
    p_meas = p,
    p_cnot = p,
    p_idle = p
)

dem = code.circuit.detector_error_model()

bm = detector_error_model_to_check_matrices(dem, allow_undecomposed_hyperedges= True)

H = bm.check_matrix.toarray()

obs = bm.observables_matrix.toarray()
```

```python
chosen_columns = np.zeros(H.shape[1], dtype = bool)
for column in range(H.shape[1]):
    if len(np.where(H[:,column] == 1)[0]) <= 3:
        chosen_columns[column] = True

H_phen = H[:, chosen_columns]
obs_phen = obs[:, chosen_columns]

    
```

```python
print(f'Distance {d}')
print('Color code cln H shape')

print(H.shape)

print('Color code sparsified cln H shape')


print(H_phen.shape)
```

```python
savemat(f"transfermatrices/ColorCodes/CC{d}.mat", {'H': H, 'obs' : obs, 'H_phen': H_phen, 'obs_phen' : obs_phen})
```
